### 1.2 Series: the DataFrame column

```python
# Series: 1D array with an index
sales = pd.Series([1200, 1500, 900, 1800, 2100],
                   index=["Jan", "Feb", "Mar", "Apr", "May"], # index labels
                   name="monthly_sales")
print(sales)
print(f"\nTotal: R$ {sales.sum():,.2f}")
print(f"Mean: R$ {sales.mean():,.2f}")
print(f"STD: R$ {sales.std():,.2f}")
```

## 3. Selection and Filtering
### 3.1 Selecting columns

```python
# One column → Series
print(df["product"])
print(type(df["product"]))
```

```python
# Multiple columns → DataFrame
df[["product", "price", "stock"]]
```

### 3.2 Indexing with `loc` and `iloc`

```python
# .loc[rows, columns] — by LABEL
print("loc — row 0, column 'product':", df.loc[0, "product"])
print()
df.loc[0:2, ["product", "price"]]
```

```python
# .iloc[rows, columns] — by POSITION (0-based, like Python)
print("iloc — row 0, column 0:", df.iloc[0, 0])
print()
df.iloc[0:3, 0:3]   # first 3 rows, first 3 columns
```

### 3.3 Boolean filtering
> This is one of the most important concepts, and it works the same way in Spark!

```python
# Simple condition
mask = df["price"] > 200
print("Boolean mask:")
print(mask.values)

# Applying the mask
df[mask] # array of True and False values.
```

```python
df[mask].reset_index(drop=True)
```

```python
# Multiple conditions
# & = AND   |  = OR   ~  = NOT

# Expensive products AND a good rating
df[(df["price"] > 200) & (df["rating"] >= 4.5)]
```

```python
# .isin() — equivalent to SQL IN
categories_of_interest = ["Peripherals", "Audio"]
df[df["category"].isin(categories_of_interest)]
```

```python
# .query() — more readable syntax (SQL-like)
df.query("price > 100 and category == 'Peripherals'")
```

### 3.4 Sorting

```python
# Sort by price (descending)
df.sort_values("price", ascending=False)
```

```python
df.sort_values("price", ascending=False).reset_index(drop=True)
```

```python
# Sort by multiple columns
df.sort_values(by=["category", "price"], ascending=[True, False])
```

```python
# Sort by multiple columns
df.sort_values(by=["price", "category"], ascending=[True, False])
```
