## 2. Introdução ao Pandas
### 2.1 Por que Pandas?

| Pandas | Apache Spark |
|--------|--------------|
| Dados em memória local | Dados distribuídos em cluster |
| `DataFrame` (linhas × colunas) | `DataFrame` (mesma ideia, escala enorme) |
| `df.groupBy()`, `df.join()` | Mesmos conceitos! |
| Ideal até ~1 GB | Terabytes / Petabytes |


> **Aprender Pandas = aprender a pensar em DataFrames** — habilidade diretamente transferível para Spark.

### 2.2 Series — a coluna do DataFrame

```python
# Series: array 1D com índice
vendas = pd.Series([1200, 1500, 900, 1800, 2100],
                   index=["Jan", "Fev", "Mar", "Abr", "Mai"], # nome das colunas
                   name="vendas_mensais")
print(vendas)
print(f"\nTotal: R$ {vendas.sum():,.2f}")
print(f"Média: R$ {vendas.mean():,.2f}")
print(f"STD: R$ {vendas.std():,.2f}")
```

## 4. Seleção e Filtragem
### 4.1 Selecionando colunas

```python
# Uma coluna → Series
print(df["produto"])
print(type(df["produto"]))
```

```python
# Múltiplas colunas → DataFrame
df[["produto", "preco", "estoque"]]
```

### 4.2 Indexação com `loc` e `iloc`

```python
# .loc[linhas, colunas] — por RÓTULO
print("loc — linha 0, coluna 'produto':", df.loc[0, "produto"])
print()
df.loc[0:2, ["produto", "preco"]]
```

```python
# .iloc[linhas, colunas] — por POSIÇÃO (0-based, como Python)
print("iloc — linha 0, coluna 0:", df.iloc[0, 0])
print()
df.iloc[0:3, 0:3]   # primeiras 3 linhas, primeiras 3 colunas
```

### 4.3 Filtragem Booleana
> Este é um dos conceitos mais importantes — funciona da mesma forma no Spark!

```python
# Condição simples
mask = df["preco"] > 200
print("Mask booleana:")
print(mask.values)

# Aplicando a máscara
df[mask] # array de True e Falses.
```

```python
df[mask].reset_index(drop=True)
```

```python
# Múltiplas condições
# & = AND   |  = OR   ~  = NOT

# Produtos caros E com boa avaliação
df[(df["preco"] > 200) & (df["avaliacao"] >= 4.5)]
```

```python
# .isin() — equivalente ao SQL IN
categorias_interesse = ["Periféricos", "Áudio"]
df[df["categoria"].isin(categorias_interesse)]
```

```python
# .query() — sintaxe mais legível (SQL-like)
df.query("preco > 100 and categoria == 'Periféricos'")
```

### 4.4 Ordenação

```python
# Ordenar por preço (decrescente)
df.sort_values("preco", ascending=False)
```

```python
df.sort_values("preco", ascending=False).reset_index(drop=True)
```

```python
# Ordenar por múltiplas colunas
df.sort_values(by=["categoria", "preco"], ascending=[True, False])
```

```python
# Ordenar por múltiplas colunas
df.sort_values(by=["preco", "categoria"], ascending=[True, False])
```
